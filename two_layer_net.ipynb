## Теоретический вопрос 1

Мы просили реализовать только ReLU, но в нейросетях можно использовать много разных функций активации, у каждой есть свои плюсы и минусы. Одна из частых проблем — исчезающие градиенты, когда во время обратного распространения градиент становится равным нулю или очень близким к нулю.

Какие из перечисленных ниже функций страдают от этой проблемы? Если рассматривать функции в одномерном случае, то для каких входных значений это происходит?

1. `sigmoid`
2. `ReLU`
3. `Leaky RELU`

### Ваш ответ:

От проблемы исчезающих или нулевых градиентов в разной степени страдают первые две функции:

1.  **Sigmoid (Сигмоида):**
    *   **Страдает ли:** Да, это основная проблема данной функции.
    *   **В каких значениях:** Это происходит при очень больших положительных или очень маленьких (больших по модулю отрицательных) входных значениях ($x \to \infty$ или $x \to -\infty$). 
    *   **Почему:** График сигмоиды в этих областях становится очень пологим («насыщается»). Поскольку производная сигмоиды вычисляется как $\sigma(x) \cdot (1 - \sigma(x))$, то когда значение функции близко к 1 или 0, производная становится практически равной нулю. При умножении таких малых чисел в глубоких сетях градиент быстро «затухает» до нуля.

2.  **ReLU:**
    *   **Страдает ли:** Да, но специфическим образом (проблема «умирающих ReLU» — *Dying ReLU*).
    *   **В каких значениях:** Это происходит для всех отрицательных входных значений ($x < 0$).
    *   **Почему:** Для любых $x < 0$ производная функции ReLU тождественно равна нулю. Если нейрон попадает в эту область, он перестает обновлять свои веса и «умирает», так как градиент через него больше не проходит. В отличие от сигмоиды, для положительных значений ($x > 0$) градиент всегда равен 1 и не исчезает.

3.  **Leaky ReLU:**
    *   **Страдает ли:** Практически нет.
    *   **Почему:** Эта функция была специально создана для решения проблемы «умирающих ReLU». При $x < 0$ она имеет небольшой наклон (обычно $0.01x$), поэтому производная равна маленькому числу (например, $0.01$), а не полному нулю. Это позволяет градиенту продолжать течь по сети, даже если входные значения отрицательны.

# Подбор гиперпараметров

**Что здесь не так?** По графикам выше видно, что функция потерь убывает почти линейно. Это часто говорит о том, что скорость обучения слишком мала. С другой стороны, слишком большая модель может начать переобучаться: тогда обучающая точность станет заметно выше валидационной. Ваша цель — подобрать такие гиперпараметры, чтобы получить качество выше `48%` на валидации.
При выборе варьируемых гиперпараметров модно отталкиваться от параметров TwoLayerNet

````python
best_model = None
best_val = -1
results = {}

hidden_sizes = [100, 150]
learning_rates = [1e-3, 2e-3]
regularization_strengths = [0.1, 0.25, 0.5]

import itertools

for hs, lr, reg in itertools.product(hidden_sizes, learning_rates, regularization_strengths):
    print(f"Testing: hidden_size={hs}, lr={lr}, reg={reg}")
    
    model = TwoLayerNet(
        input_dim=input_size, 
        hidden_dim=hs, 
        num_classes=10, 
        reg=reg,
        weight_scale=1e-3
    )
    
    solver = Solver(
        model, data,
        update_rule='sgd',
        optim_config={'learning_rate': lr},
        lr_decay=0.95,
        num_epochs=5, 
        batch_size=200,
        verbose=False 
    )
    
    solver.train()
    
    val_acc = solver.best_val_acc
    results[(hs, lr, reg)] = val_acc
    print(f"Validation accuracy: {val_acc:.4f}")
    
    if val_acc > best_val:
        best_val = val_acc
        best_model = model

print(f"\nBest validation accuracy achieved: {best_val:.4f}")
````

## Теоретический вопрос 2

После обучения нейросетевого классификатора может оказаться, что точность на тесте заметно ниже, чем на обучении. Как можно уменьшить этот разрыв? Выберите все подходящие варианты.

1. Обучать на более крупном датасете.
2. Добавить больше скрытых нейронов.
3. Увеличить силу регуляризации.
4. Ничего из перечисленного.

### Ваш ответ

1.  **Обучать на более крупном датасете (Вариант 1):** Это один из самых эффективных способов борьбы с переобучением. Чем больше разнообразных примеров видит модель, тем сложнее ей подстроиться под конкретные шумы и тем лучше она выявляет общие закономерности. Это напрямую уменьшает разрыв между обучением и тестом.
2.  **Увеличить силу регуляризации (Вариант 3):** Регуляризация (например, L2-weight decay) накладывает штраф за слишком большие веса. Это ограничивает сложность модели, не давая ей слишком сильно подстраиваться под тренировочные данные. Это «сглаживает» модель и заставляет её находить более общие признаки.

### Почему остальные варианты не подходят:
*   **Добавить больше скрытых нейронов (Вариант 2):** Это действие увеличивает «ёмкость» (мощность) модели. Если модель и так переобучена, добавление новых нейронов только **усугубит** проблему: у сети появится ещё больше ресурсов, чтобы выучить тренировочный датасет наизусть, и разрыв между точностью на обучении и тесте только вырастет.
*   **Ничего из перечисленного (Вариант 4):** Неверно, так как 1 и 3 являются классическими методами борьбы с переобучением.
